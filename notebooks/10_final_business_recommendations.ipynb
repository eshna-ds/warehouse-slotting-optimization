```python
import pandas as pd

final_recommendations = pd.read_csv(
    "../data/processed/final_slot_recommendations.csv"
)

final_recommendations.head()
```

```python
top_products = final_recommendations.sort_values(
    "picking_priority",
    ascending=False
)

top_products[
    [
        "StockCode",
        "cluster_name",
        "recommended_zone",
        "picking_priority",
        "distance_saved"
    ]
].head(20)
```

```python
zone_summary = final_recommendations.groupby(
    "recommended_zone"
).agg(
    product_count=("StockCode", "count"),
    avg_priority=("picking_priority", "mean"),
    avg_distance_saved=("distance_saved", "mean")
)

zone_summary
```

```python
cluster_summary_final = final_recommendations.groupby(
    "cluster_name"
).agg(
    product_count=("StockCode", "count"),
    avg_order_frequency=("order_frequency", "mean"),
    avg_total_quantity=("total_quantity", "mean"),
    avg_distance_saved=("distance_saved", "mean")
)

cluster_summary_final
```

```python
highest_savings = final_recommendations.sort_values(
    "distance_saved",
    ascending=False
)

highest_savings[
    [
        "StockCode",
        "cluster_name",
        "recommended_zone",
        "distance_saved"
    ]
].head(20)
```

```python
final_recommendations["business_recommendation"] = (
    "Place in " +
    final_recommendations["recommended_zone"] +
    " based on product movement and picking priority."
)
```

```python
final_recommendations[
    [
        "StockCode",
        "cluster_name",
        "recommended_zone",
        "picking_priority",
        "business_recommendation"
    ]
].head()
```

```python
final_recommendations.to_csv(
    "../data/processed/business_slotting_recommendations.csv",
    index=False
)
```
